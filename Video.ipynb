```python
path = "Images_-60_0_-135"
files = os.listdir(path)
files.sort()
frames = np.zeros((len(files),1300,1300,3), dtype=np.uint8)

for i, f in enumerate(files):
    im_path = os.path.join(path,f)
    im = cv2.imread(im_path)
    frames[i] = np.asarray(im)[:1300, 600:1900]
```

```python
# write video
h, w = frames[0].shape[:2]
fps = 5
out = cv2.VideoWriter("Growth.mp4", cv2.VideoWriter_fourcc(*"mp4v"), fps, (w, h))

for fr in frames:
    out.write(fr)

out.release()
```
