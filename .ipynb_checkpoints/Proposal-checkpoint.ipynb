$\textbf{1) Having a multistrategy EGT along a dormancy spectrum.}$

For example:
    - x_1: healthy cells
    - x_2: single cancer cells
    - x_3: small cancer clusters
    - x_4: dormant micrometastases

These phenotypes would respond differently to the fluctuating environment
    
$\textbf{2) More realistic environmental feedback (activation).}$

Instead of $f(n) = \frac{n^\nu}{k+n^\nu}$, with $\nu=1$, I would variate the value of $\nu$.

Also use other sigmoidal response functions, such as logistic or Gompertz
    
$\textbf{3) Specific gene expression mechanisms for the strategies $x_i$} $

Depending on the advancement stage of the cell (strategy), they could have evolved/developed different gene expression strategies, that allow them to interact differently with the environment. (Klein 2020)

$\textbf{4) Time varying toxin production}$

Couple $\epsilon$ to gene expression.


From now on this is advancing to agent-based models

$\textbf{5) Invasion waves and environmental preconditioning}$

Implement waves of invasion of cancer cells, and study how the alteration of the environment would facilitate further invasions. Also, study how the degradation rate of the toxin $\mu$ (or regeneration factor) influence these subsequent invasions.

$\textbf{6) Redistribution of resources}$

Model environmental dynamics as:

$\frac{\partial n}{\partial t} = F(n, \vec{x}) + D_n\nabla^2n$ 
following Lin 2019 in order capture transport and local depletion.

$\textbf{7) Multiscale models}$

Embed the dormancy dynamics in bone architectures. Use multimodal data (light sheet and micro/nanoCT) to parameterize models.
